```python

with open("api-key.txt", "r") as f:
    key = f.readline()

from openai import OpenAI
import pandas as pd

client = OpenAI(api_key= key, base_url="https://api.deepseek.com")
root_dir = "1_data_preprocessing\dataset\culture_wvs"

china_df = pd.read_csv("1_data_preprocessing\dataset\culture_wvs\majority_answers_CHN_50pct_sum_5pct_close.csv")
question= china_df["QuestionText"].iloc[0]
answer = china_df["AnswerText"].iloc[0]

messages = [{ "role": "user", "content": " You will generate dialogue between two persons using the data found in 1_data_preprocessing\dataset\culture_wvs\majority_answers_CHN_50pct_sum_5pct_close.csv and output it using JSON file.\n Example:  "  }]

response = client.chat.completions.create(
    model="deepseek-chat",
    messages= messages,
    max_tokens=1024,
    temperature=0.7,
    stream=False
)

print(response.choices[0].message.content)
```
